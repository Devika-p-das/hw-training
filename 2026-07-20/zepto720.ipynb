```python
store_oos = (
    df.pivot_table(
        index=["city", "store_id"],
        columns="out_of_stock_flag",
        aggfunc="size",
        fill_value=0
    )
    .reset_index()
)

# Ensure both columns exist
for col in ["Yes", "No"]:
    if col not in store_oos.columns:
        store_oos[col] = 0

store_oos["Total"] = store_oos["Yes"] + store_oos["No"]
store_oos["Out_of_stock_%"] = (
    store_oos["Yes"] / store_oos["Total"] * 100
).round(2)

store_oos = store_oos.sort_values(
    "Out_of_stock_%",
    ascending=False
)

store_oos
```

```python
import pandas as pd

# Convert price columns to numeric
price_cols = [
    "item_mrp",
    "item_selling_price",
    "discount_in_percent"
]

for col in price_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Normalize discount flag
flag = df["discount_flag"].astype(str).str.strip().str.lower()

# 1. Selling Price > MRP
selling_gt_mrp = df[df["item_selling_price"] > df["item_mrp"]]

print(f"1. Selling Price > MRP : {len(selling_gt_mrp)} rows")
print(selling_gt_mrp[[
    "unique_id",
    "product_name",
    "item_mrp",
    "item_selling_price"
]])

# 2. Discount Flag = Yes but MRP == Selling Price
discount_no_difference = df[
    (flag == "yes") &
    (df["item_mrp"] == df["item_selling_price"])
]

print(f"\n2. Discount Flag = Yes but MRP == Selling Price : {len(discount_no_difference)} rows")
print(discount_no_difference[[
    "unique_id",
    "product_name",
    "item_mrp",
    "item_selling_price",
    "discount_flag"
]])

# 3. Discount % > 100
discount_gt_100 = df[df["discount_in_percent"] > 100]

print(f"\n3. Discount % > 100 : {len(discount_gt_100)} rows")
print(discount_gt_100[[
    "unique_id",
    "product_name",
    "item_mrp",
    "item_selling_price",
    "discount_in_percent"
]])
```

```python
import pandas as pd

# Set your threshold
threshold = 150

# Products with long names
long_products = df[df["product_name"].str.len() > threshold].copy()

print(f"Products with name length > {threshold}: {len(long_products)}")
print(f"Unique URLs: {long_products['product_url'].nunique()}")
```

```python
import requests
from tqdm import tqdm

not_found = []

for idx, url in tqdm(long_products["product_url"].dropna().items()):
    try:
        r = requests.get(
            url,
            timeout=10,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        if r.status_code == 404:
            not_found.append(idx)

    except Exception:
        pass

df_404 = long_products.loc[not_found]

print(f"404 URLs among long-name products: {len(df_404)}")

df_404[["unique_id", "product_name", "product_url"]]
```

```python
import requests
from tqdm import tqdm
from collections import Counter

status_counts = Counter()

for url in tqdm(long_products["product_url"].dropna()):
    try:
        r = requests.get(
            url,
            timeout=10,
            headers={"User-Agent": "Mozilla/5.0"}
        )
        status_counts[r.status_code] += 1
    except Exception:
        status_counts["ERROR"] += 1

print(status_counts)
```

```python
import asyncio
from playwright.async_api import async_playwright

url = "https://www.zepto.com/pn/fastup-reload-ready-to-drink-low-sugar-energy-drink-instant-energy-hydration-drink-lemon-flavour-500ml-jimmys-sex-on-the-beach-cocktail-mixer-250ml-combo/pvid/faa49fa5-6ff0-4321-9f66-ac40569e44e2"

async def check_url():
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()

        response = await page.goto(url, wait_until="networkidle")

        print("HTTP Status:", response.status if response else "No response")
        print("Final URL:", page.url)
        print("Title:", await page.title())

        content = (await page.content()).lower()

        if "404" in content or "page not found" in content or "product not found" in content:
            print("❌ Broken product page")
        else:
            print("✅ Product page exists")

        await browser.close()

await check_url()
```

```python
import re

# Normalize brand names
df["brand_norm"] = (
    df["brand_name"]
    .fillna("")
    .str.lower()
    .str.replace(r"[^a-z0-9]", "", regex=True)   # remove spaces, ', -, &, etc.
)

# Find brands that become the same after normalization
brand_issue = (
    df.groupby("brand_norm")
    .agg(
        variant_count=("brand_name", "nunique"),
        brand_variants=("brand_name", lambda x: sorted(set(x)))
    )
    .reset_index()
)

brand_issue = brand_issue[brand_issue["variant_count"] > 1]

brand_issue
```
