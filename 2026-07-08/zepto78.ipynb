```python
dept_category = (
    df.groupby("department")["category"]
    .apply(lambda x: sorted(x.unique()))
    .reset_index()
)

dept_category
```

```python
same_product = (
    df.groupby(["product_name","product_id_upc_ean", "product_grammage", "net_quantity"])
    .agg(
        url_count=( "product_url", "nunique"),
       product_urls=( "product_url", lambda x: list(x.unique()))
    )
    .reset_index()
)

same_product = same_product[same_product["url_count"] > 1]

same_product
```

```python
master = df[["department", "category"]].drop_duplicates()
city_stock = (
    df.groupby("city")["out_of_stock_flag"]
    .value_counts()
    .unstack(fill_value=0)
)

city_stock["Total"] = city_stock.sum(axis=1)
city_stock["Out_of_stock_%"] = (
    city_stock["Yes"] / city_stock["Total"] * 100
).round(2)

city_stock.sort_values("Out_of_stock_%", ascending=False)
```

```python
dept_stock = (
    df.groupby("department")["out_of_stock_flag"]
    .value_counts()
    .unstack(fill_value=0)
)

dept_stock["Total"] = dept_stock.sum(axis=1)
dept_stock["Out_of_stock_%"] = (
    dept_stock["Yes"] / dept_stock["Total"] * 100
).round(2)

dept_stock.sort_values("Out_of_stock_%", ascending=False)
```

```python
cat_stock = (
    df.groupby("category")["out_of_stock_flag"]
    .value_counts()
    .unstack(fill_value=0)
)

cat_stock["Total"] = cat_stock.sum(axis=1)
cat_stock["Out_of_stock_%"] = (
    cat_stock["Yes"] / cat_stock["Total"] * 100
).round(2)

cat_stock.sort_values("Out_of_stock_%", ascending=False)
```

```python
same_store_diff = (
    df.groupby(["store_id", "store_name_location", "city", "unique_id"])
    .agg(
        product_name=("product_name", "first"),
        price_count=("item_selling_price", "nunique"),
        discount_count=("discount_in_percent", "nunique"),
        mrp_count=("item_mrp", "nunique")
    )
    .reset_index()
)

same_store_diff = same_store_diff[
    (same_store_diff["price_count"] > 1) |
    (same_store_diff["discount_count"] > 1) |
    (same_store_diff["mrp_count"] > 1)
]

same_store_diff
```

```python
import re

# Normalize brand names
df["brand_norm"] = (
    df["brand_name"]
    .fillna("")
    .str.lower()
    .str.replace(r"[^a-z0-9]", "", regex=True)   # remove spaces, ', -, &, etc.
)

# Find brands that become the same after normalization
brand_issue = (
    df.groupby("brand_norm")
    .agg(
        variant_count=("brand_name", "nunique"),
        brand_variants=("brand_name", lambda x: sorted(set(x)))
    )
    .reset_index()
)

brand_issue = brand_issue[brand_issue["variant_count"] > 1]

brand_issue
```

```python
productid_name_issue = (
    df.groupby("product_id_upc_ean")
    .agg(
        product_name_count=("product_name", "nunique"),
        product_names=("product_name", lambda x: sorted(x.unique()))
    )
    .reset_index()
)

productid_name_issue = productid_name_issue[
    productid_name_issue["product_name_count"] > 1
]

productid_name_issue
```

```python
store_oos = (
    df.pivot_table(
        index=["city", "store_id"],
        columns="out_of_stock_flag",
        aggfunc="size",
        fill_value=0
    )
    .reset_index()
)

# Ensure both columns exist
for col in ["Yes", "No"]:
    if col not in store_oos.columns:
        store_oos[col] = 0

store_oos["Total"] = store_oos["Yes"] + store_oos["No"]
store_oos["Out_of_stock_%"] = (
    store_oos["Yes"] / store_oos["Total"] * 100
).round(2)

store_oos = store_oos.sort_values(
    "Out_of_stock_%",
    ascending=False
)

store_oos
```
