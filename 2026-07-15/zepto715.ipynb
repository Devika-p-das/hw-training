```python
import re

# Normalize brand names
df["productname_norm"] = (
    df["product_name"]
    .fillna("")
    .str.lower()
    .str.replace(r"[^a-z0-9]", "", regex=True)   # remove spaces, ', -, &, etc.
)

# Find brands that become the same after normalization
name_issue = (
    df.groupby("productname_norm")
    .agg(
        variant_count=("product_name", "nunique"),
        name_variants=("product_name", lambda x: sorted(set(x)))
    )
    .reset_index()
)

name_issue = name_issue[name_issue["variant_count"] > 1]

name_issue
```

```python
# 1. Group by identity columns to isolate the exact same physical product
product_identity_issues = (
    df.groupby(["unique_id", "product_grammage", "net_quantity"])
    .agg(
        # Count how many different spelling/name variations exist for this product
        name_variant_count=("product_name", "nunique"),
        # Create an alphabetised list of those unique name variations
        name_variants=("product_name", lambda x: sorted(set(x))),
        # Optional: capture variations in brand name for the same product
        brand_variants=("brand_name", lambda x: sorted(set(x.dropna()))),
    )
    .reset_index()
)

# 2. Filter for rows where the same product has more than 1 name variation
name_issues_by_id = product_identity_issues[
    product_identity_issues["name_variant_count"] > 1
]

# 3. View the final flagged discrepancies
print(name_issues_by_id)
```

```python
# 1. Group by unique_id to see if physical attributes split
id_integrity_check = (
    df.groupby("unique_id")
    .agg(
        grammage_count=("product_grammage", "nunique"),
        grammage_variants=("product_grammage", lambda x: list(set(x.dropna()))),
        quantity_count=("net_quantity", "nunique"),
        quantity_variants=("net_quantity", lambda x: list(set(x.dropna()))),
        category_count=("category", "nunique"),
        category_variants=("category", lambda x: list(set(x.dropna()))),
    )
    .reset_index()
)

# 2. Filter for rows where an ID maps to MORE THAN ONE physical variation
corrupted_ids = id_integrity_check[
    (id_integrity_check["grammage_count"] > 1) | 
    (id_integrity_check["quantity_count"] > 1)
]

# 3. Display the errors
print(f"Found {len(corrupted_ids)} unique_ids tied to different products.")
print(corrupted_ids)
```
