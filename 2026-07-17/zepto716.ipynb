```python
store_oos = (
    df.pivot_table(
        index=["city", "store_id"],
        columns="out_of_stock_flag",
        aggfunc="size",
        fill_value=0
    )
    .reset_index()
)

# Ensure both columns exist
for col in ["Yes", "No"]:
    if col not in store_oos.columns:
        store_oos[col] = 0

store_oos["Total"] = store_oos["Yes"] + store_oos["No"]
store_oos["Out_of_stock_%"] = (
    store_oos["Yes"] / store_oos["Total"] * 100
).round(2)

store_oos = store_oos.sort_values(
    "Out_of_stock_%",
    ascending=False
)

store_oos
```

```python
import re

# Normalize brand names
df["brand_norm"] = (
    df["brand_name"]
    .fillna("")
    .str.lower()
    .str.replace(r"[^a-z0-9]", "", regex=True)   # remove spaces, ', -, &, etc.
)

# Find brands that become the same after normalization
brand_issue = (
    df.groupby("brand_norm")
    .agg(
        variant_count=("brand_name", "nunique"),
        brand_variants=("brand_name", lambda x: sorted(set(x)))
    )
    .reset_index()
)

brand_issue = brand_issue[brand_issue["variant_count"] > 1]

brand_issue
```

```python
# 1. Group by unique_id to see if physical attributes split
id_integrity_check = (
    df.groupby("unique_id")
    .agg(
        grammage_count=("product_grammage", "nunique"),
        grammage_variants=("product_grammage", lambda x: list(set(x.dropna()))),
        quantity_count=("net_quantity", "nunique"),
        quantity_variants=("net_quantity", lambda x: list(set(x.dropna()))),
        category_count=("category", "nunique"),
        category_variants=("category", lambda x: list(set(x.dropna()))),
    )
    .reset_index()
)

# 2. Filter for rows where an ID maps to MORE THAN ONE physical variation
corrupted_ids = id_integrity_check[
    (id_integrity_check["grammage_count"] > 1) | 
    (id_integrity_check["quantity_count"] > 1)
]

# 3. Display the errors
print(f"Found {len(corrupted_ids)} unique_ids tied to different products.")
print(corrupted_ids)
```

```python
# 1. Group by identity columns to isolate the exact same physical product
product_identity_issues = (
    df.groupby(["unique_id", "product_grammage", "net_quantity"])
    .agg(
        # Count how many different spelling/name variations exist for this product
        name_variant_count=("product_name", "nunique"),
        # Create an alphabetised list of those unique name variations
        name_variants=("product_name", lambda x: sorted(set(x))),
        # Optional: capture variations in brand name for the same product
        brand_variants=("brand_name", lambda x: sorted(set(x.dropna()))),
    )
    .reset_index()
)

# 2. Filter for rows where the same product has more than 1 name variation
name_issues_by_id = product_identity_issues[
    product_identity_issues["name_variant_count"] > 1
]

# 3. View the final flagged discrepancies
print(name_issues_by_id)
```

```python
import pandas as pd

# Convert price columns to numeric
price_cols = [
    "item_mrp",
    "item_selling_price",
    "discount_in_percent"
]

for col in price_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Normalize discount flag
flag = df["discount_flag"].astype(str).str.strip().str.lower()

# 1. Selling Price > MRP
selling_gt_mrp = df[df["item_selling_price"] > df["item_mrp"]]

print(f"1. Selling Price > MRP : {len(selling_gt_mrp)} rows")
print(selling_gt_mrp[[
    "unique_id",
    "product_name",
    "item_mrp",
    "item_selling_price"
]])

# 2. Discount Flag = Yes but MRP == Selling Price
discount_no_difference = df[
    (flag == "yes") &
    (df["item_mrp"] == df["item_selling_price"])
]

print(f"\n2. Discount Flag = Yes but MRP == Selling Price : {len(discount_no_difference)} rows")
print(discount_no_difference[[
    "unique_id",
    "product_name",
    "item_mrp",
    "item_selling_price",
    "discount_flag"
]])

# 3. Discount % > 100
discount_gt_100 = df[df["discount_in_percent"] > 100]

print(f"\n3. Discount % > 100 : {len(discount_gt_100)} rows")
print(discount_gt_100[[
    "unique_id",
    "product_name",
    "item_mrp",
    "item_selling_price",
    "discount_in_percent"
]])
```

```python
import pandas as pd
import requests
from concurrent.futures import ThreadPoolExecutor, as_completed
from tqdm import tqdm

headers = {
    "User-Agent": "Mozilla/5.0"
}

unique_urls = df["product_url"].dropna().unique()

def check_url(url):
    try:
        r = requests.get(
            url,
            headers=headers,
            timeout=10,
            allow_redirects=True
        )
        return url, r.status_code
    except Exception:
        return url, "ERROR"

results = []

with ThreadPoolExecutor(max_workers=20) as executor:
    futures = [executor.submit(check_url, url) for url in unique_urls]

    for future in tqdm(as_completed(futures), total=len(futures)):
        results.append(future.result())

status_df = pd.DataFrame(results, columns=["product_url", "status_code"])

invalid_urls = status_df[status_df["status_code"] != 200]

print(invalid_urls)
```

```python
import pandas as pd
import requests
import time
from concurrent.futures import ThreadPoolExecutor, as_completed
from tqdm import tqdm

headers = {
    "User-Agent": "Mozilla/5.0"
}

# Sample 500 unique URLs
sample_urls = (
    pd.Series(df["product_url"].dropna().unique())
    .sample(500, random_state=42)
    .tolist()
)

def check_url(url):
    try:
        time.sleep(0.5)  # Reduce request rate
        r = requests.get(
            url,
            headers=headers,
            timeout=10,
            allow_redirects=True,
        )
        return url, r.status_code
    except Exception:
        return url, "ERROR"

results = []

with ThreadPoolExecutor(max_workers=3) as executor:
    futures = [executor.submit(check_url, url) for url in sample_urls]

    for future in tqdm(as_completed(futures), total=len(futures)):
        results.append(future.result())

status_df = pd.DataFrame(results, columns=["product_url", "status_code"])

print(status_df["status_code"].value_counts())
```
