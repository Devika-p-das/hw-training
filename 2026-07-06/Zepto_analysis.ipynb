### Data cleaning

```python
# Removing the rows which are completely empty(product_type,product_description ,package_type,unit_price,product_image_urls

# Remove columns that are completely empty
data = data.dropna(axis=1, how='all')
```

#### Unique_id check

```python
product_name_check = (
    data.groupby("unique_id")["product_name"]
    .nunique()
)

product_name_check[product_name_check > 1]
```

```python
# Find unique_ids with multiple product names
anomaly_ids = (
    data.groupby("unique_id")["product_name"]
    .nunique()
    .loc[lambda x: x > 1]
    .index
)

# Fetch rows for those unique_ids and remove duplicate product names
result = (
    data[data["unique_id"].isin(anomaly_ids)]
    .drop_duplicates(subset=["unique_id", "product_name"])
    .sort_values(["unique_id", "product_name"])
)

result
```

```python
store_locations = [
    "Anika Redwood Square, Nagarabhavi Main Rd, 2nd Block, Coconut Garden, Chalukya Nagar, Bhyraveshwara Nagar, Bengaluru, Karnataka, 560072, India",
    "Anju Boutique Cosmetic, 920, Jaina Market, Nathuram Market, DLF Phase 4, Nathuram Market, Chakkarpur, Sector 28, Gurugram, Haryana, 122002, India"
]

monitoring = (
    data[data["store_name_location"].isin(store_locations)]
    [["store_id", "store_name_location", "product_url"]]
    .drop_duplicates()
    .sort_values(["store_name_location", "store_id"])
)

monitoring
```

```python
monitoring_table = (
    data.groupby("store_name_location")
    .agg(
        number_of_store_ids=("store_id", "nunique"),
        store_ids=("store_id", lambda x: ", ".join(sorted(x.unique())))
    )
    .reset_index()
)

# Keep only locations with more than one store_id
monitoring_table = monitoring_table[
    monitoring_table["number_of_store_ids"] > 1
]

monitoring_table
```
