# Two-Week Product Data Comparison
**Objective:** Compare product assortment, pricing, availability, categories, brands, ratings/reviews, and data quality between two weekly extracts of the same e-commerce competitor.

**Inputs:** two full-dump CSV exports (Week 1 = older file, Week 2 = newer file).

Sections:
1. Setup & Config
2. Load & Inspect Data
3. Cleaning & Standardisation
4. Product Key Matching (Additions / Removals / Continuing)
5. KPI Comparison
6. Price Analysis
7. Category & Brand Analysis
8. Stock / Availability Analysis
9. Rating & Review Analysis
10. Data Quality Assessment
11. Additional Insights
12. Dashboard
13. Executive Summary


```python
# ============================================================
# 1. LOAD DATA (pipe-delimited)
# ============================================================

import pandas as pd
import numpy as np
import pandas as pd

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option("display.max_colwidth", None)

OLD_FILE = "/home/user/Downloads/DataHut_SI_Dm_FullDump_20260827.CSV"
NEW_FILE = "/home/user/Downloads/DataHut_SI_Dm_FullDump_20260903.CSV"

def safe_read(path):
    return pd.read_csv(
        path,
        dtype=str,
        engine="python",
        sep="|",                 # <-- pipe separator
        quotechar='"',
        on_bad_lines="warn",
        encoding="utf-8",
        encoding_errors="replace",
    )

df_old = safe_read(OLD_FILE)
df_new = safe_read(NEW_FILE)

df_old.columns = [c.strip() for c in df_old.columns]
df_new.columns = [c.strip() for c in df_new.columns]

print("Week1 shape:", df_old.shape)
print("Week2 shape:", df_new.shape)
print(df_old.columns.tolist())
```

```python
# Quick structural check - confirm both files share the same schema
missing_in_new = set(df_old.columns) - set(df_new.columns)
missing_in_old = set(df_new.columns) - set(df_old.columns)
print("Columns in Week1 but not Week2:", missing_in_new)
print("Columns in Week2 but not Week1:", missing_in_old)
```

```python
# ============================================================
# 3. CLEAN KEY FIELDS
# ============================================================

for df in (df_old, df_new):
    for c in ["unique_id", "product_name", "pdp_url"]:
        if c in df.columns:
            df[c] = df[c].astype(str).str.strip()
            df.loc[df[c].isin(["", "nan", "None"]), c] = np.nan

print("Week1 unique_id nulls:", df_old["unique_id"].isna().sum())
print("Week2 unique_id nulls:", df_new["unique_id"].isna().sum())
```

```python
# ============================================================
# NULL VALUE COMPARISON — Week1 vs Week2, all columns side by side
# ============================================================

import pandas as pd
import numpy as np

def null_report(df):
    n = len(df)
    return pd.DataFrame({
        "Column": df.columns,
        "Null_Count": df.isna().sum().values,
        "Null_%": (df.isna().sum().values / n * 100).round(2),
    })

null_w1 = null_report(df_old)
null_w2 = null_report(df_new)

comparison = pd.merge(
    null_w1.rename(columns={"Null_Count": "Null_Count_W1", "Null_%": "Null_%_W1"}),
    null_w2.rename(columns={"Null_Count": "Null_Count_W2", "Null_%": "Null_%_W2"}),
    on="Column", how="outer"
)

comparison["Null_%_Diff (W2-W1)"] = comparison["Null_%_W2"] - comparison["Null_%_W1"]
comparison = comparison.sort_values("Null_%_W2", ascending=False).reset_index(drop=True)

pd.set_option("display.max_rows", 130)
display(comparison)

comparison.to_csv("null_comparison_all_columns.csv", index=False)
print("\nSaved: null_comparison_all_columns.csv")
```

```python
# ============================================================
# 4. PRODUCT-LEVEL ANALYSIS: additions, missing, common
# ============================================================

# de-duplicate on unique_id within each week (keep last occurrence)
df_old_u = df_old.dropna(subset=["unique_id"]).drop_duplicates(subset=["unique_id"], keep="last")
df_new_u = df_new.dropna(subset=["unique_id"]).drop_duplicates(subset=["unique_id"], keep="last")

old_ids = set(df_old_u["unique_id"])
new_ids = set(df_new_u["unique_id"])

added_ids = new_ids - old_ids       # new in Week2
missing_ids = old_ids - new_ids     # dropped from Week1
common_ids = old_ids & new_ids      # present in both

print("================ PRODUCT COUNT SUMMARY ================")
print(f"Week1 total unique products: {len(old_ids)}")
print(f"Week2 total unique products: {len(new_ids)}")
print(f"Additional products (new in Week2):    {len(added_ids)}")
print(f"Missing products (dropped from Week1): {len(missing_ids)}")
print(f"Common products (in both weeks):       {len(common_ids)}")

added_df = df_new_u[df_new_u["unique_id"].isin(added_ids)][["unique_id", "product_name", "pdp_url"]].reset_index(drop=True)
missing_df = df_old_u[df_old_u["unique_id"].isin(missing_ids)][["unique_id", "product_name", "pdp_url"]].reset_index(drop=True)
common_df = df_new_u[df_new_u["unique_id"].isin(common_ids)][["unique_id", "product_name", "pdp_url"]].reset_index(drop=True)

print(f"\nAdditional products: {len(added_df)}")
display(added_df)

print(f"\nMissing products: {len(missing_df)}")
display(missing_df)

print(f"\nCommon products: {len(common_df)}")
display(common_df)

added_df.to_csv("additional_products.csv", index=False)
missing_df.to_csv("missing_products.csv", index=False)
common_df.to_csv("common_products.csv", index=False)
print("\nSaved: additional_products.csv, missing_products.csv, common_products.csv")
```

```python
# ============================================================
# PRICE & PROMOTION COMPARISON — common products (unique_id) across weeks
# ============================================================

import pandas as pd
import numpy as np

PRICE_COLS = ["regular_price", "selling_price", "price_was", "promotion_price", "percentage_discount"]
PROMO_COLS = ["promotion_type", "promotion_valid_from", "promotion_valid_upto", "promotion_description"]

# --- clean numeric price columns ---
for df in (df_old, df_new):
    for c in PRICE_COLS:
        if c in df.columns:
            df[c] = pd.to_numeric(df[c], errors="coerce")
    for c in PROMO_COLS:
        if c in df.columns:
            df[c] = df[c].astype(str).str.strip()
            df.loc[df[c].isin(["", "nan", "None"]), c] = np.nan

# --- de-dup on unique_id (already have df_old_u / df_new_u from before, but redo safely) ---
df_old_u = df_old.dropna(subset=["unique_id"]).drop_duplicates(subset=["unique_id"], keep="last")
df_new_u = df_new.dropna(subset=["unique_id"]).drop_duplicates(subset=["unique_id"], keep="last")

cols_needed = ["unique_id", "product_name", "pdp_url"] + PRICE_COLS + PROMO_COLS
cols_needed = [c for c in cols_needed if c in df_old_u.columns]

merged = pd.merge(
    df_old_u[cols_needed],
    df_new_u[cols_needed],
    on="unique_id", suffixes=("_w1", "_w2"), how="inner"
)

print("Common products compared for pricing:", len(merged))
```

```python
# ============================================================
# 1. SELLING PRICE CHANGES
# ============================================================

merged["selling_price_delta"] = merged["selling_price_w2"] - merged["selling_price_w1"]
merged["selling_price_pct_change"] = (merged["selling_price_delta"] / merged["selling_price_w1"]) * 100

price_changed = merged[merged["selling_price_delta"].fillna(0) != 0]

print(f"Products with selling_price CHANGED: {len(price_changed)} / {len(merged)} ({len(price_changed)/len(merged):.1%})")
print(f"  - Increased: {(merged['selling_price_delta'] > 0).sum()}")
print(f"  - Decreased: {(merged['selling_price_delta'] < 0).sum()}")
print(f"  - Unchanged: {(merged['selling_price_delta'] == 0).sum()}")

display(price_changed[["unique_id", "product_name_w2", "selling_price_w1", "selling_price_w2",
                        "selling_price_delta", "selling_price_pct_change"]]
        .sort_values("selling_price_delta", ascending=False)
        .head(20))
```

```python
# ============================================================
# 2. REGULAR PRICE CHANGES (base price before any promo)
# ============================================================

merged["regular_price_delta"] = merged["regular_price_w2"] - merged["regular_price_w1"]
regular_changed = merged[merged["regular_price_delta"].fillna(0) != 0]

print(f"Products with regular_price CHANGED: {len(regular_changed)} / {len(merged)}")
display(regular_changed[["unique_id", "product_name_w2", "regular_price_w1", "regular_price_w2", "regular_price_delta"]]
        .sort_values("regular_price_delta", ascending=False)
        .head(20))
```

```python
# ============================================================
# 3. PROMOTION STATUS CHANGES
# ============================================================

merged["on_promo_w1"] = merged["promotion_price_w1"].notna()
merged["on_promo_w2"] = merged["promotion_price_w2"].notna()

newly_on_promo = merged[(~merged["on_promo_w1"]) & (merged["on_promo_w2"])]
newly_off_promo = merged[(merged["on_promo_w1"]) & (~merged["on_promo_w2"])]
promo_both_weeks = merged[(merged["on_promo_w1"]) & (merged["on_promo_w2"])]

print(f"On promotion Week1: {merged['on_promo_w1'].sum()} | Week2: {merged['on_promo_w2'].sum()}")
print(f"Newly went ON promotion (W1→W2): {len(newly_on_promo)}")
print(f"Came OFF promotion (W1→W2): {len(newly_off_promo)}")
print(f"On promotion in BOTH weeks: {len(promo_both_weeks)}")

print("\nSample - newly on promotion:")
display(newly_on_promo[["unique_id", "product_name_w2", "selling_price_w1", "selling_price_w2",
                         "promotion_price_w2", "percentage_discount_w2"]].head(10))

print("\nSample - came off promotion:")
display(newly_off_promo[["unique_id", "product_name_w2", "selling_price_w1", "selling_price_w2",
                          "promotion_price_w1", "percentage_discount_w1"]].head(10))
```

```python
# ============================================================
# 4. PROMOTION PRICE / DISCOUNT % CHANGES (for products on promo in both weeks)
# ============================================================

promo_both_weeks = promo_both_weeks.copy()
promo_both_weeks["promo_price_delta"] = promo_both_weeks["promotion_price_w2"] - promo_both_weeks["promotion_price_w1"]
promo_both_weeks["discount_pct_delta"] = promo_both_weeks["percentage_discount_w2"] - promo_both_weeks["percentage_discount_w1"]

promo_changed = promo_both_weeks[promo_both_weeks["promo_price_delta"].fillna(0) != 0]
print(f"Products with promotion_price CHANGED (while on promo both weeks): {len(promo_changed)}")
display(promo_changed[["unique_id", "product_name_w2", "promotion_price_w1", "promotion_price_w2",
                        "promo_price_delta", "percentage_discount_w1", "percentage_discount_w2", "discount_pct_delta"]]
        .sort_values("promo_price_delta")
        .head(20))
```

```python
# ============================================================
# 5. DATA QUALITY FLAGS ON PRICING
# ============================================================

# selling_price higher than regular_price (shouldn't normally happen)
bad_w1 = merged[merged["selling_price_w1"] > merged["regular_price_w1"]]
bad_w2 = merged[merged["selling_price_w2"] > merged["regular_price_w2"]]
print(f"selling_price > regular_price — Week1: {len(bad_w1)}, Week2: {len(bad_w2)}")

# promotion_price present but no discount %, or vice versa
promo_no_discount_w2 = merged[merged["promotion_price_w2"].notna() & merged["percentage_discount_w2"].isna()]
print(f"promotion_price present but percentage_discount missing (Week2): {len(promo_no_discount_w2)}")

# negative or zero prices
neg_price_w2 = merged[merged["selling_price_w2"] <= 0]
print(f"selling_price <= 0 (Week2): {len(neg_price_w2)}")

# promotion_price higher than selling_price (illogical — promo should be cheaper)
promo_higher_w2 = merged[merged["promotion_price_w2"] > merged["selling_price_w2"]]
print(f"promotion_price > selling_price (Week2, illogical): {len(promo_higher_w2)}")
```

```python
# ============================================================
# 6. SUMMARY TABLE (for report)
# ============================================================

summary = pd.DataFrame({
    "Metric": [
        "Common products compared",
        "Selling price changed",
        "Selling price increased",
        "Selling price decreased",
        "Regular price changed",
        "Newly went on promotion",
        "Came off promotion",
        "On promotion both weeks",
        "Promotion price changed (both weeks on promo)",
    ],
    "Count": [
        len(merged),
        len(price_changed),
        (merged["selling_price_delta"] > 0).sum(),
        (merged["selling_price_delta"] < 0).sum(),
        len(regular_changed),
        len(newly_on_promo),
        len(newly_off_promo),
        len(promo_both_weeks),
        len(promo_changed),
    ]
})
display(summary)

merged.to_csv("price_promotion_comparison.csv", index=False)
print("\nSaved: price_promotion_comparison.csv")
```

```python
# ============================================================
# CATEGORY & BRAND ANALYSIS — Week1 vs Week2
# ============================================================

import pandas as pd
import numpy as np

CAT_COLS = ["producthierarchy_level1", "producthierarchy_level2", "producthierarchy_level3"]

for df in (df_old, df_new):
    for c in CAT_COLS + ["brand", "brand_type"]:
        if c in df.columns:
            df[c] = df[c].astype(str).str.strip().str.title()
            df.loc[df[c].isin(["", "Nan", "None"]), c] = np.nan

# use the de-duped (unique_id) tables from earlier
# df_old_u, df_new_u already exist
```

```python
# ============================================================
# 1. CATEGORY (Level 1) DISTRIBUTION — counts & change
# ============================================================

def dist_change(col, df_a, df_b, top_n=20):
    a = df_a[col].value_counts(dropna=True)
    b = df_b[col].value_counts(dropna=True)
    out = pd.DataFrame({"Week1_Count": a, "Week2_Count": b}).fillna(0)
    out["Change"] = out["Week2_Count"] - out["Week1_Count"]
    out["% Change"] = np.where(out["Week1_Count"] > 0, out["Change"] / out["Week1_Count"] * 100, np.nan)
    return out.sort_values("Week2_Count", ascending=False).head(top_n)

cat_change = dist_change("producthierarchy_level1", df_old_u, df_new_u)
print("Category (Level 1) distribution — Week1 vs Week2:")
display(cat_change.round(1))
```

```python
# ============================================================
# 2. CATEGORY (Level 2) — more granular view
# ============================================================

cat2_change = dist_change("producthierarchy_level2", df_old_u, df_new_u)
print("Category (Level 2) distribution — Week1 vs Week2:")
display(cat2_change.round(1))
```

```python
# ============================================================
# 3. BRAND DISTRIBUTION — counts & change
# ============================================================

brand_change = dist_change("brand", df_old_u, df_new_u, top_n=25)
print("Brand distribution — Week1 vs Week2 (top 25 by Week2 volume):")
display(brand_change.round(1))
```

```python
# ============================================================
# 4. NEW / DROPPED CATEGORIES & BRANDS (entirely absent in one week)
# ============================================================

old_cats = set(df_old_u["producthierarchy_level1"].dropna())
new_cats = set(df_new_u["producthierarchy_level1"].dropna())
print("Categories new to Week2:", new_cats - old_cats)
print("Categories dropped after Week1:", old_cats - new_cats)

old_brands = set(df_old_u["brand"].dropna())
new_brands = set(df_new_u["brand"].dropna())
new_entrant_brands = new_brands - old_brands
dropped_brands = old_brands - new_brands

print(f"\nBrands new to Week2: {len(new_entrant_brands)}")
print(sorted(new_entrant_brands)[:30])
print(f"\nBrands dropped after Week1: {len(dropped_brands)}")
print(sorted(dropped_brands)[:30])
```

```python
# ============================================================
# 5. TOP GAINERS / LOSERS BY PRODUCT COUNT — category & brand
# ============================================================

print("Top 10 categories by ABSOLUTE product count increase:")
display(cat_change.sort_values("Change", ascending=False).head(10))

print("\nTop 10 categories by ABSOLUTE product count decrease:")
display(cat_change.sort_values("Change").head(10))

print("\nTop 10 brands by ABSOLUTE product count increase:")
display(brand_change.sort_values("Change", ascending=False).head(10))

print("\nTop 10 brands by ABSOLUTE product count decrease:")
display(brand_change.sort_values("Change").head(10))
```

```python
# ============================================================
# 6. BRAND CONCENTRATION — market share within catalogue
# ============================================================

def top_share(df, col, n=10):
    counts = df[col].value_counts(dropna=True)
    share = (counts / counts.sum() * 100).round(2)
    return pd.DataFrame({"Product_Count": counts, "% of Catalogue": share}).head(n)

print("Top 10 brands by catalogue share — Week1:")
display(top_share(df_old_u, "brand"))

print("\nTop 10 brands by catalogue share — Week2:")
display(top_share(df_new_u, "brand"))
```

```python
# ============================================================
# STOCK / AVAILABILITY ANALYSIS
# ============================================================

for df in (df_old, df_new, df_old_u, df_new_u):
    if "instock" in df.columns:
        df["instock_bool"] = df["instock"].astype(str).str.strip().str.lower().map({
            "true": True, "yes": True, "1": True, "in stock": True,
            "false": False, "no": False, "0": False, "out of stock": False
        })

stock_rate_w1 = df_old_u["instock_bool"].mean() * 100
stock_rate_w2 = df_new_u["instock_bool"].mean() * 100
print(f"Overall in-stock rate — Week1: {stock_rate_w1:.1f}%  |  Week2: {stock_rate_w2:.1f}%")
print(f"Change: {stock_rate_w2 - stock_rate_w1:+.1f} percentage points")
```

```python
# ============================================================
# 10. EXPORT FOR REPORT
# ============================================================

cat_change.to_csv("category_change.csv")
cat2_change.to_csv("category_level2_change.csv")
brand_change.to_csv("brand_change.csv")

print("Saved: category_change.csv, category_level2_change.csv, brand_change.csv,")
print("       newly_out_of_stock.csv, newly_back_in_stock.csv")
```

```python
# ============================================================
# RATING & REVIEW COMPARISON — products common to BOTH weeks
# ============================================================

import pandas as pd
import numpy as np

for df in (df_old_u, df_new_u):
    df["rating"] = pd.to_numeric(df["rating"], errors="coerce")
    df["review"] = pd.to_numeric(df["review"], errors="coerce")

rr_cols = ["unique_id", "product_name", "brand", "producthierarchy_level1", "rating", "review", "pdp_url"]
rr_cols = [c for c in rr_cols if c in df_old_u.columns]

# --- merge only common products (inner join = present in both weeks) ---
common_rr = pd.merge(
    df_old_u[rr_cols],
    df_new_u[["unique_id", "rating", "review"]],
    on="unique_id", suffixes=("_w1", "_w2"), how="inner"
)

print("Total common products:", len(common_rr))
```

```python
# ============================================================
# 1. Restrict to products with a rating in BOTH weeks
# ============================================================

both_rated = common_rr.dropna(subset=["rating_w1", "rating_w2"]).copy()
both_rated["rating_delta"] = both_rated["rating_w2"] - both_rated["rating_w1"]
both_rated["review_delta"] = both_rated["review_w2"] - both_rated["review_w1"]

print(f"Common products with a rating in BOTH weeks: {len(both_rated)} / {len(common_rr)}")
print(f"  - Rating increased: {(both_rated['rating_delta'] > 0).sum()}")
print(f"  - Rating decreased: {(both_rated['rating_delta'] < 0).sum()}")
print(f"  - Rating unchanged: {(both_rated['rating_delta'] == 0).sum()}")

print(f"\nAvg rating Week1: {both_rated['rating_w1'].mean():.2f} | Week2: {both_rated['rating_w2'].mean():.2f}")
print(f"Avg review count Week1: {both_rated['review_w1'].mean():.1f} | Week2: {both_rated['review_w2'].mean():.1f}")
print(f"Total review count Week1: {both_rated['review_w1'].sum():,.0f} | Week2: {both_rated['review_w2'].sum():,.0f}")
```

```python
# ============================================================
# 2. Biggest rating changes
# ============================================================

print("Top 10 rating INCREASES:")
display(both_rated.sort_values("rating_delta", ascending=False).head(10)
        [["unique_id", "product_name", "brand", "rating_w1", "rating_w2", "rating_delta", "review_w1", "review_w2"]])

print("\nTop 10 rating DECREASES:")
display(both_rated.sort_values("rating_delta").head(10)
        [["unique_id", "product_name", "brand", "rating_w1", "rating_w2", "rating_delta", "review_w1", "review_w2"]])
```

```python
# ============================================================
# 4. Summary table for report (self-contained)
# ============================================================

review_shrunk = both_rated[both_rated["review_delta"] < 0]

summary = pd.DataFrame({
    "Metric": [
        "Common products (both weeks)",
        "Common products rated in both weeks",
        "Rating increased",
        "Rating decreased",
        "Rating unchanged",
        "Avg rating - Week1",
        "Avg rating - Week2",
        "Total reviews - Week1",
        "Total reviews - Week2",
        "Review count decreased (flag)",
    ],
    "Value": [
        len(common_rr),
        len(both_rated),
        (both_rated["rating_delta"] > 0).sum(),
        (both_rated["rating_delta"] < 0).sum(),
        (both_rated["rating_delta"] == 0).sum(),
        round(both_rated["rating_w1"].mean(), 2),
        round(both_rated["rating_w2"].mean(), 2),
        int(both_rated["review_w1"].sum()),
        int(both_rated["review_w2"].sum()),
        len(review_shrunk),
    ]
})
display(summary)

both_rated.to_csv("common_products_rating_review_comparison.csv", index=False)
print("\nSaved: common_products_rating_review_comparison.csv")
```

```python
# ============================================================
# KPI COMPARISON — Week1 vs Week2
# ============================================================

import pandas as pd
import numpy as np

# ensure numeric types are correct
for df in (df_old_u, df_new_u):
    for c in ["regular_price", "selling_price", "price_was", "promotion_price", "percentage_discount", "rating", "review"]:
        if c in df.columns:
            df[c] = pd.to_numeric(df[c], errors="coerce")
    if "instock" in df.columns and "instock_bool" not in df.columns:
        df["instock_bool"] = df["instock"].astype(str).str.strip().str.lower().map({
            "true": True, "yes": True, "1": True, "in stock": True,
            "false": False, "no": False, "0": False, "out of stock": False
        })

def kpi_row(df, label):
    return {
        "Week": label,
        "Total Unique Products": df["unique_id"].nunique(),
        "Unique Brands": df["brand"].nunique() if "brand" in df else np.nan,
        "Unique Categories (L1)": df["producthierarchy_level1"].nunique() if "producthierarchy_level1" in df else np.nan,
        "Avg Selling Price": df["selling_price"].mean(),
        "Median Selling Price": df["selling_price"].median(),
        "% On Promotion": df["promotion_price"].notna().mean() * 100 if "promotion_price" in df else np.nan,
        "Avg Discount % (on promo)": df.loc[df["promotion_price"].notna(), "percentage_discount"].mean() if "percentage_discount" in df else np.nan,
        "% In Stock": df["instock_bool"].mean() * 100 if "instock_bool" in df else np.nan,
        "Products with a Rating": df["rating"].notna().sum() if "rating" in df else np.nan,
        "Avg Rating": df["rating"].mean() if "rating" in df else np.nan,
        "Total Review Count": df["review"].sum() if "review" in df else np.nan,
    }

kpi_df = pd.DataFrame([
    kpi_row(df_old_u, "Week1 (2026-08-27)"),
    kpi_row(df_new_u, "Week2 (2026-09-03)"),
]).set_index("Week").T

kpi_df["Absolute Change"] = kpi_df.iloc[:, 1] - kpi_df.iloc[:, 0]
kpi_df["% Change"] = (kpi_df["Absolute Change"] / kpi_df.iloc[:, 0]) * 100

display(kpi_df.round(2))

kpi_df.to_csv("kpi_comparison.csv")
print("\nSaved: kpi_comparison.csv")
```

```python
# ============================================================
# 1. INTERACTIVE SCATTER — Price change vs Rating change (hover = product)
#    Lets you spot products with a suspicious combo (e.g. big price hike + rating drop)
# ============================================================

cols = ["unique_id", "product_name", "brand", "producthierarchy_level1", "selling_price", "rating", "review"]
cols = [c for c in cols if c in df_old_u.columns]

combo = pd.merge(
    df_old_u[cols], df_new_u[cols],
    on="unique_id", suffixes=("_w1", "_w2")
).dropna(subset=["selling_price_w1", "selling_price_w2", "rating_w1", "rating_w2"])

combo["price_pct_change"] = (combo["selling_price_w2"] - combo["selling_price_w1"]) / combo["selling_price_w1"] * 100
combo["rating_delta"] = combo["rating_w2"] - combo["rating_w1"]

fig = px.scatter(
    combo, x="price_pct_change", y="rating_delta",
    color="producthierarchy_level1_w2",
    hover_data=["product_name_w2", "brand_w2", "selling_price_w1", "selling_price_w2", "rating_w1", "rating_w2"],
    title="Price % Change vs Rating Change (hover for product details)",
    labels={"price_pct_change": "Selling Price % Change", "rating_delta": "Rating Change",
            "producthierarchy_level1_w2": "Category"},
)
fig.add_hline(y=0, line_dash="dash", line_color="gray")
fig.add_vline(x=0, line_dash="dash", line_color="gray")
fig.update_layout(height=600)
fig.show()
```

```python
# ============================================================
# 2. INTERACTIVE TREEMAP — Category > Brand hierarchy, sized by product count
#    Click into a category to drill down into its brands. Toggle week via dropdown.
# ============================================================

def build_treemap_df(df, label):
    t = df.dropna(subset=["producthierarchy_level1", "brand"]).copy()
    t["week"] = label
    return t

tm_data = pd.concat([
    build_treemap_df(df_old_u, WEEK1_LABEL),
    build_treemap_df(df_new_u, WEEK2_LABEL),
])

fig = px.treemap(
    tm_data, path=["week", "producthierarchy_level1", "brand"],
    title="Category → Brand Treemap (product count) — Week1 vs Week2",
)
fig.update_traces(textinfo="label+value+percent parent")
fig.update_layout(height=650)
fig.show()
```

```python
# ============================================================
# 4. INTERACTIVE STOCK HEATMAP — Out-of-stock rate by Category x Brand (Week2)
#    Reveals which brand+category combos are most stock-troubled
# ============================================================

if "instock" in df_new_u.columns:
    df_new_u["instock_bool"] = df_new_u["instock"].astype(str).str.lower().map({
        "true": True, "yes": True, "1": True, "in stock": True,
        "false": False, "no": False, "0": False, "out of stock": False
    })

oos_pivot = df_new_u.dropna(subset=["producthierarchy_level1", "brand"]).copy()
top_cats = oos_pivot["producthierarchy_level1"].value_counts().head(10).index
top_brands = oos_pivot["brand"].value_counts().head(15).index
oos_pivot = oos_pivot[oos_pivot["producthierarchy_level1"].isin(top_cats) & oos_pivot["brand"].isin(top_brands)]

pivot = oos_pivot.pivot_table(
    index="brand", columns="producthierarchy_level1", values="instock_bool",
    aggfunc=lambda s: (~s.fillna(True)).mean() * 100
)

fig = px.imshow(
    pivot, text_auto=".0f", aspect="auto", color_continuous_scale="Reds",
    title="Out-of-Stock Rate (%) — Top Brands x Top Categories (Week2)",
    labels=dict(color="OOS %")
)
fig.update_layout(height=600)
fig.show()
```

```python
# ============================================================
# EMPTY / NULL VALUE CHECK — every column, both weeks
# ============================================================

import pandas as pd
import numpy as np

def null_report(df, label):
    n = len(df)
    report = pd.DataFrame({
        "Column": df.columns,
        "Null_Count": df.isna().sum().values,
        "Null_%": (df.isna().sum().values / n * 100).round(2),
        "Non_Null_Count": df.notna().sum().values,
    })
    report["Week"] = label
    return report.sort_values("Null_%", ascending=False)

null_w1 = null_report(df_old, "Week1 (2026-08-27)")
null_w2 = null_report(df_new, "Week2 (2026-09-03)")

print(f"Week1 total rows: {len(df_old)}")
print(f"Week2 total rows: {len(df_new)}")
```

```python
# ============================================================
# 2. Columns that are COMPLETELY EMPTY (100% null) — in either week
# ============================================================

fully_empty_w1 = null_w1[null_w1["Null_%"] == 100]["Column"].tolist()
fully_empty_w2 = null_w2[null_w2["Null_%"] == 100]["Column"].tolist()

print(f"Fully empty columns — Week1 ({len(fully_empty_w1)}):")
print(fully_empty_w1)

print(f"\nFully empty columns — Week2 ({len(fully_empty_w2)}):")
print(fully_empty_w2)

print(f"\nFully empty in BOTH weeks: {sorted(set(fully_empty_w1) & set(fully_empty_w2))}")
```

```python
# ============================================================
# CORRELATION BETWEEN RATING AND REVIEW COUNT — Week1 & Week2
# ============================================================

import pandas as pd
import numpy as np
from scipy import stats

for df in (df_old_u, df_new_u):
    df["rating"] = pd.to_numeric(df["rating"], errors="coerce")
    df["review"] = pd.to_numeric(df["review"], errors="coerce")
```

```python
# ============================================================
# 1. Pearson & Spearman correlation — each week independently
# ============================================================

def corr_report(df, label):
    d = df.dropna(subset=["rating", "review"])
    pearson_r, pearson_p = stats.pearsonr(d["rating"], d["review"])
    spearman_r, spearman_p = stats.spearmanr(d["rating"], d["review"])
    print(f"--- {label} ---")
    print(f"n = {len(d)}")
    print(f"Pearson r  = {pearson_r:.3f}  (p={pearson_p:.4f})")
    print(f"Spearman ρ = {spearman_r:.3f}  (p={spearman_p:.4f})\n")
    return pearson_r, spearman_r

r_w1, s_w1 = corr_report(df_old_u, "Week1")
r_w2, s_w2 = corr_report(df_new_u, "Week2")
```

```python
# ============================================================
# 2. Correlation on the LOG of review count
#    (review counts are usually heavily right-skewed — a few products with
#     thousands of reviews dominate a raw Pearson correlation; log-transform
#     gives a more meaningful linear relationship)
# ============================================================

for df in (df_old_u, df_new_u):
    df["log_review"] = np.log1p(df["review"])

def corr_report_log(df, label):
    d = df.dropna(subset=["rating", "log_review"])
    pearson_r, pearson_p = stats.pearsonr(d["rating"], d["log_review"])
    print(f"--- {label} (rating vs log(review+1)) ---")
    print(f"Pearson r = {pearson_r:.3f}  (p={pearson_p:.4f})\n")
    return pearson_r

r_log_w1 = corr_report_log(df_old_u, "Week1")
r_log_w2 = corr_report_log(df_new_u, "Week2")
```

```python
# ============================================================
# 3. VISUALIZATION — scatter with regression line, both weeks side by side
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))

for ax, df, label, r in zip(axes, [df_old_u, df_new_u], ["Week1", "Week2"], [r_w1, r_w2]):
    d = df.dropna(subset=["rating", "review"])
    sns.regplot(data=d, x="rating", y="review", ax=ax, scatter_kws={"alpha": 0.3, "s": 15}, line_kws={"color": "red"})
    ax.set_title(f"{label}: Rating vs Review Count\nPearson r = {r:.3f}")
    ax.set_xlabel("Rating")
    ax.set_ylabel("Review Count")

plt.tight_layout()
plt.savefig("rating_review_correlation_scatter.png", dpi=150)
plt.show()
```

```python
# ============================================================
# 4. VISUALIZATION — using log(review) since raw review count is skewed
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))

for ax, df, label, r in zip(axes, [df_old_u, df_new_u], ["Week1", "Week2"], [r_log_w1, r_log_w2]):
    d = df.dropna(subset=["rating", "log_review"])
    sns.regplot(data=d, x="rating", y="log_review", ax=ax, scatter_kws={"alpha": 0.3, "s": 15}, line_kws={"color": "red"})
    ax.set_title(f"{label}: Rating vs log(Review Count + 1)\nPearson r = {r:.3f}")
    ax.set_xlabel("Rating")
    ax.set_ylabel("log(Review Count + 1)")

plt.tight_layout()
plt.savefig("rating_log_review_correlation_scatter.png", dpi=150)
plt.show()
```

```python
# ============================================================
# DATA QUALITY ASSESSMENT — comprehensive rule-based checks
# ============================================================

import pandas as pd
import numpy as np

for df in (df_old_u, df_new_u):
    for c in ["selling_price", "regular_price", "promotion_price", "percentage_discount", "rating", "review"]:
        if c in df.columns:
            df[c] = pd.to_numeric(df[c], errors="coerce")

def dq_checks(df, label):
    n = len(df)
    checks = {}

    checks["Total rows"] = n
    checks["Duplicate unique_id"] = df["unique_id"].duplicated().sum()
    checks["Missing product_name"] = df["product_name"].isna().sum()
    checks["Missing brand"] = df["brand"].isna().sum()
    checks["Missing category (L1)"] = df["producthierarchy_level1"].isna().sum()
    checks["Missing selling_price"] = df["selling_price"].isna().sum()
    checks["selling_price <= 0"] = (df["selling_price"] <= 0).sum()
    checks["selling_price > regular_price"] = (df["selling_price"] > df["regular_price"]).sum()
    checks["Missing pdp_url"] = df["pdp_url"].isna().sum() if "pdp_url" in df.columns else np.nan
    checks["Malformed pdp_url"] = (df["pdp_url"].notna() & ~df["pdp_url"].str.startswith(("http://", "https://"))).sum() if "pdp_url" in df.columns else np.nan
    checks["Rating out of 0-5 range"] = df["rating"].apply(lambda x: pd.notna(x) and not (0 <= x <= 5)).sum()
    checks["Has rating but no review count"] = (df["rating"].notna() & df["review"].isna()).sum()
    checks["Has review count but no rating"] = (df["review"].notna() & df["rating"].isna()).sum()
    checks["Missing barcode"] = df["barcode"].isna().sum() if "barcode" in df.columns else np.nan

    return pd.Series(checks, name=label)

dq_w1 = dq_checks(df_old_u, "Week1")
dq_w2 = dq_checks(df_new_u, "Week2")

dq_table = pd.concat([dq_w1, dq_w2], axis=1)
dq_table["Diff (W2-W1)"] = dq_table["Week2"] - dq_table["Week1"]

display(dq_table)
dq_table.to_csv("data_quality_assessment.csv")
print("Saved: data_quality_assessment.csv")
```

```python
# ============================================================
# ADDITIONAL INSIGHT 1 — Assortment churn rate & category-level churn
# ============================================================

old_ids = set(df_old_u["unique_id"])
new_ids = set(df_new_u["unique_id"])
added_ids = new_ids - old_ids
missing_ids = old_ids - new_ids

churn_rate = (len(added_ids) + len(missing_ids)) / len(old_ids | new_ids) * 100
print(f"Insight 1 — Overall assortment churn rate: {churn_rate:.2f}%")
print(f"  ({len(added_ids)} added + {len(missing_ids)} removed) out of {len(old_ids | new_ids)} total unique products")

# which categories saw the most churn (added+removed combined)
added_df = df_new_u[df_new_u["unique_id"].isin(added_ids)]
missing_df = df_old_u[df_old_u["unique_id"].isin(missing_ids)]

churn_by_cat = pd.concat([
    added_df["producthierarchy_level1"].value_counts().rename("Added"),
    missing_df["producthierarchy_level1"].value_counts().rename("Removed"),
], axis=1).fillna(0)
churn_by_cat["Total_Churn"] = churn_by_cat["Added"] + churn_by_cat["Removed"]
churn_by_cat = churn_by_cat.sort_values("Total_Churn", ascending=False)

print("\nTop 10 categories by churn (added + removed):")
display(churn_by_cat.head(10))
```

```python
# ============================================================
# ADDITIONAL INSIGHT 2 — Highest-rated products that were REMOVED
#                        (potential loss of best-sellers)
# ============================================================

if "rating" in missing_df.columns:
    lost_highrated = missing_df.dropna(subset=["rating"]).sort_values("rating", ascending=False).head(10)
    print("Insight 2 — Highest-rated products removed between weeks:")
    display(lost_highrated[["unique_id", "product_name", "brand", "rating", "review", "selling_price", "pdp_url"]])
```

```python
# ============================================================
# ADDITIONAL INSIGHT 3 — Brands with the most volatile pricing
# ============================================================

price_cols = ["unique_id", "brand", "selling_price"]
price_matched = pd.merge(
    df_old_u[price_cols], df_new_u[["unique_id", "selling_price"]],
    on="unique_id", suffixes=("_w1", "_w2")
)
price_matched["pct_change"] = (price_matched["selling_price_w2"] - price_matched["selling_price_w1"]) / price_matched["selling_price_w1"] * 100

brand_volatility = price_matched.dropna(subset=["pct_change"]).groupby("brand")["pct_change"].agg(["mean", "std", "count"])
brand_volatility = brand_volatility[brand_volatility["count"] >= 5].sort_values("std", ascending=False)

print("Insight 3 — Brands with most volatile pricing (std dev of % change, min 5 products):")
display(brand_volatility.round(2).head(10))
```

```python
# ============================================================
# ADDITIONAL INSIGHT 4 — Products with rating/review inconsistent with price change
#                        (e.g. rating dropped WHILE price increased — risk signal)
# ============================================================

combo = pd.merge(
    df_old_u[["unique_id", "product_name", "brand", "selling_price", "rating"]],
    df_new_u[["unique_id", "selling_price", "rating"]],
    on="unique_id", suffixes=("_w1", "_w2")
).dropna(subset=["selling_price_w1", "selling_price_w2", "rating_w1", "rating_w2"])

combo["price_pct_change"] = (combo["selling_price_w2"] - combo["selling_price_w1"]) / combo["selling_price_w1"] * 100
combo["rating_delta"] = combo["rating_w2"] - combo["rating_w1"]

risk_flag = combo[(combo["price_pct_change"] > 5) & (combo["rating_delta"] < 0)]
print(f"Insight 4 — Products with price increase (>5%) AND rating decrease: {len(risk_flag)}")
display(risk_flag.sort_values("price_pct_change", ascending=False).head(10)
        [["unique_id", "product_name", "brand", "selling_price_w1", "selling_price_w2", "price_pct_change", "rating_w1", "rating_w2", "rating_delta"]])
```

```python
# ============================================================
# WHICH CATEGORY (producthierarchy_level1) DECREASED PRICE THE MOST
# ============================================================

import pandas as pd
import numpy as np

for df in (df_old_u, df_new_u):
    df["selling_price"] = pd.to_numeric(df["selling_price"], errors="coerce")

cols = ["unique_id", "producthierarchy_level2", "selling_price"]
matched = pd.merge(
    df_old_u[cols], df_new_u[["unique_id", "selling_price"]],
    on="unique_id", suffixes=("_w1", "_w2")
).dropna(subset=["selling_price_w1", "selling_price_w2"])

matched["price_delta"] = matched["selling_price_w2"] - matched["selling_price_w1"]
matched["price_pct_change"] = (matched["price_delta"] / matched["selling_price_w1"]) * 100

# only products that actually had a price decrease
decreased = matched[matched["price_delta"] < 0]

cat_price_decrease = decreased.groupby("producthierarchy_level2").agg(
    n_products_decreased=("unique_id", "count"),
    avg_pct_decrease=("price_pct_change", "mean"),
    avg_abs_decrease=("price_delta", "mean"),
    total_abs_decrease=("price_delta", "sum"),
).sort_values("avg_pct_decrease")  # most negative % first = biggest average decrease

print("Categories ranked by AVERAGE % price decrease (steepest discount first):")
display(cat_price_decrease.round(2))
```

```python
# ============================================================
# Also rank by NUMBER of products that decreased (most affected category)
# ============================================================

cat_by_count = cat_price_decrease.sort_values("n_products_decreased", ascending=False)
print("Categories ranked by NUMBER of products with a price decrease:")
display(cat_by_count.round(2))
```
